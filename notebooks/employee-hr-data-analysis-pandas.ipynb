```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": [" Ali ", "Sara", "Ahmed", "Usman", "Ayesha", "Bilal", "Hina", "Omar"],
    "Age": [25, 29, None, 35, 28, 31, 26, 40],
    "Gender": ["M", "Female", "Male", "M", "F", "Male", "Female", "M"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "IT"],
    "City": ["Islamabad", "Lahore", " Islamabad ", "Karachi", "Lahore", "ISLAMABAD", "Karachi", "Lahore"],
    "Joining_Date": ["2022-01-15", "2021-03-20", "2023-06-10", "2020-08-12",
                     "2022-11-05", "2021-07-18", "2023-01-25", "2019-04-30"],
    "Salary": [75000, 65000, "N/A", 90000, 62000, 85000, 58000, 100000],
    "Bonus": [5000, 4000, 6000, None, 3500, 5000, 3000, 8000],
    "Performance": [85, 72, 90, 65, None, 88, 70, 95],
    "Status": ["Active", "Active", "Active", "Inactive", "Active", "Active", "Active", "Inactive"]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# Clean the Gender column
df["Gender"]=df["Gender"].replace({
    "F":"Female",
    "M":"Male"
})
```

```python
# Clean the City column
df["City"]=df["City"].str.strip().str.title()
```

```python
# Salary
df["Salary"]=pd.to_numeric(df["Salary"],errors="coerce")
```

```python
# Bonus
df["Bonus"].isna().sum()
```

```python
# What should we do with the missing Bonus?
df["Bonus"]=df["Bonus"].fillna(df["Bonus"].mean())
```
